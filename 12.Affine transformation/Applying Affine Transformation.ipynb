<p style="font-size:24px"> Shearing an image with Scikit-image </p>

```python
from skimage.io import imread
from scipy.ndimage import affine_transform
import matplotlib.pyplot as plt
img = imread('../images-videos/lenna.jpg')

def shearing_img(img, sh_x=0, sh_y=0):
    shear_matrix = np.array([[1, -sh_x, 0], [-sh_y, 1, 0], [0, 0, 1]])
    return affine_transform(img, shear_matrix)

sheared_image_x = shearing_img(img, 0.2, 0)
sheared_image_y = shearing_img(img, 0, 0.3)
sheared_image_xy = shearing_img(img, 0.2, 0.3)


plt.figure(figsize=(5, 5))

plt.subplot(221)
plt.imshow(img)
plt.title('Original Image', size=14)
plt.axis('off')

plt.subplot(222)
plt.imshow(sheared_image_x)
plt.title('Sheared Image-x', size=14)
plt.axis('off')

plt.subplot(223)
plt.imshow(sheared_image_y)
plt.title('Sheard Image-y', size=14)
plt.axis('off')

plt.subplot(224)
plt.imshow(sheared_image_xy)
plt.title('Sheard Image-xy', size=14)
plt.axis('off')

plt.show()
```

<p style="font-size:24px"> Shearing an image with OpenCV </p>

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('../images-videos/lenna.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

def shearing_img(img, sh_x=0, sh_y=0):
    shear_matrix = np.array([[1, sh_x, 0], [sh_y, 1, 0]])
    sheared_img = cv2.warpAffine(img, shear_matrix, (img.shape[1], img.shape[0]))
    return sheared_img

sheared_x = shearing_img(img, 0, 0.2)
sheared_y = shearing_img(img, 0.4, 0)
shear_matrix_xy = shearing_img(img, 0.4, 0.2)

plt.figure(figsize=(5, 5))

plt.subplot(221)
plt.imshow(img)
plt.title('Original Image', size=14)
plt.axis('off')

plt.subplot(222)
plt.imshow(sheared_x)
plt.title('Sheared Image-x', size=14)
plt.axis('off')

plt.subplot(223)
plt.imshow(sheared_y)
plt.title('Sheared Image-y', size=14)
plt.axis('off')

plt.subplot(224)
plt.imshow(shear_matrix_xy)
plt.title('Sheard Image-xy', size=14)
plt.axis('off')

plt.show()
```

<p style="font-size:24px"> Translating an image with Scikit-image </p>

```python
from skimage.io import imread
from skimage.transform import AffineTransform, warp
import matplotlib.pyplot as plt

img = imread('../images-videos/lenna.jpg')

def translating_img(img, shift_x=0, shift_y=0):
    translate_matrix = AffineTransform(translation=(shift_y, shift_x))
    return warp(img, translate_matrix.inverse)

shift_x =150
shift_y = 90

translated_x = translating_img(img, shift_x, 0)
translated_y = translating_img(img, 0, shift_y)
translated_xy = translating_img(img, shift_x, shift_y)

plt.figure(figsize=(5, 5))

plt.subplot(221)
plt.imshow(img)
plt.title('Original Image', size=12)
plt.axis('off')

plt.subplot(222)
plt.imshow(translated_x)
plt.title('Shiftted Image in x', size=12)
plt.axis('off')

plt.subplot(223)
plt.imshow(translated_y)
plt.title('Shiftted Image in y', size=12)
plt.axis('off')

plt.subplot(224)
plt.imshow(translated_xy)
plt.title('Shiftted Image in xy', size=12)
plt.axis('off')

plt.show()
```

<p style="font-size:24px"> Translating an image with OpenCV </p>

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

img = cv2.imread('../images-videos/lenna.jpg')
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

def translate_img(img, shift_x=0, shift_y=0):
    translation_matrix = np.array([[1, 0, shift_x], [0, 1, shift_y]], dtype=np.float32)
    return cv2.warpAffine(img, translation_matrix, (img.shape[1], img.shape[0]))

shift_x = 170
shift_y = 70

translated_x = translate_img(img, shift_x, 0)
translated_y = translate_img(img, 0, shift_y)
translated_xy = translate_img(img, shift_x, shift_y)

plt.figure(figsize=(5, 5))

plt.subplot(221)
plt.imshow(img)
plt.title('Original Image', size=12)
plt.axis('off')

plt.subplot(222)
plt.imshow(translated_x)
plt.title('Shiftted Image in x', size=12)
plt.axis('off')

plt.subplot(223)
plt.imshow(translated_y)
plt.title('Shiftted Image in y', size=12)
plt.axis('off')

plt.subplot(224)
plt.imshow(translated_xy)
plt.title('Shiftted Image in xy', size=12)
plt.axis('off')

plt.show()
```
