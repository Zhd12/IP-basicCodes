<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">The FFT with the numpy.fft module</p>

 <div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">The DFT of an image can be computed with the numpy.fft module's similar set of functions.</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Computing the magnitude and phase of a DFT</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">We'll use the house.jpg image as input and thus fft2() to get the real and imaginary components of the Fourier coefficients<br>
After that, we'll compute the magnitude/spectrum and the phase and, finally, use ifft2() to reconstruct the image:</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Import libraries</p>

```python
import numpy as np
from skimage.io import imread
from skimage.color import rgb2gray
import matplotlib.pyplot as plt
```

<p style="font-family:Bookman Old Style;font-size:27px">Reading and Converting Image to Grayscale</p>

```python
img = rgb2gray(imread('../images-videos/house.jpg'))
```

<p style="font-family:Bookman Old Style;font-size:27px">Applying FFT and Shifting the Frequency</p>

```python
freq_img = np.fft.fft2(img)
shift_img = np.fft.fftshift(freq_img)
```

<p style="font-family:Bookman Old Style;font-size:27px">Calculating Magnitude and Phase of the Frequency Spectrum</p>

```python
magnitude_img = 20 * (np.log10(np.abs(shift_img)) + 0.01)
phase_img = np.angle(shift_img)
```

<p style="font-family:Bookman Old Style;font-size:27px">Reconstructing the Image Using Inverse FFT</p>

```python
real_freq_img = np.fft.ifft2(freq_img).real
```

<p style="font-family:Bookman Old Style;font-size:27px">Plotting Results</p>

```python
plt.figure(figsize=(12, 6))

plt.subplot(221)
plt.imshow(img, cmap='gray')
plt.title('Original Image', size=17)
plt.axis('off')

plt.subplot(222)
plt.imshow(magnitude_img, cmap='gray')
plt.title('FFT Spectrum Magnitude', size=17)
plt.axis('off')

plt.subplot(223)
plt.imshow(phase_img, cmap='gray')
plt.title('FFT Phase', size=17)
plt.axis('off')

plt.subplot(224)
plt.imshow(real_freq_img, cmap='gray')
plt.title('Reconstructed Image', size=17)
plt.axis('off')
plt.show()
```
