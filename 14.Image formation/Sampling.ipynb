<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Sampling</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">Sampling refers to the selection/rejection of image pixels, which means that it is a spatial operation. We can use sampling to increase or reduce the size of an image, with up-sampling and down-sampling, respectively.</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">UP-Sampling</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">in order to increase the size of an image we need to up-sample the image. The challenge is that the new larger images will have some pixels that have no corresponding pixel in the original smaller image, and we need to guess those unknown pixel values.<br> We can guess the value of an
 unknown pixel using the following:<br><br>
 1) An aggregate, for example, the mean value of its nearest known one or more pixel-neighbors values <br><br>2) An interpolated value using pixel-neighbors with bi-linear or cubic interpolation</p></div>

<p style="font-family:Bookman Old Style;font-size:24px">1.Nearest neighbor-based up-sampling</p>

```python
from PIL import Image
import matplotlib.pyplot as plt
from show_multi_images import show_images, show_single_img

img = Image.open("../images-videos/lenna-100.png")
plt.imshow(img)
# guess the unknown pixel with nearest known one
resized_img = img.resize((img.width*3, img.height*3), Image.NEAREST)
plt.figure(figsize=(10, 10))
plt.imshow(resized_img)
plt.show()
```

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">the output image is pixelated (with the blocking artifacts and jagged edges) and is of poor
 quality.<br>
 for smooth this artifacts use interpolation method
</p></div>

<p style="font-family:Bookman Old Style;font-size:24px">2.Up-sampling and interpolation</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">In order to improve the up-sampled output image quality, we can use some interpolation method such as bi-linear or bi-cubic interpolation.</p></div>

<p style="font-family:Bookman Old Style;font-size:24px">2.1.Bi-linear interpolation</p>

```python
from PIL import Image
import matplotlib.pyplot as plt

img = Image.open("../images-videos/lenna-100.png")
plt.imshow(img)
# guess the unknown pixel with nearest known one
resized_img = img.resize((img.width*3, img.height*3), Image.BILINEAR)
plt.figure(figsize=(10, 10))
plt.imshow(resized_img)
plt.show()
```

<p style="font-family:Bookman Old Style;font-size:27px">DOWN-Sampling</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">In order to decrease the size of an image, we need to down-sample the image. For each pixel in the new smaller image, there will be multiple pixels in the original larger image. We can compute the value of a pixel in the new image by doing the following:<br><br>
 - Dropping some pixels (for example, dropping every other row and column if we
 want an image a fourth of the size of the original image) from the larger image in
 a systematic way <br><br>
 - Computing the new pixel value as an aggregate value of the corresponding
 multiple pixels in the original image</p></div>

```python
from PIL import Image
import matplotlib.pyplot as plt
im = Image.open("../images-videos/lenna.jpg")
plt.imshow(im), plt.show()
im1 = im.resize((im.width//5, im.height//5), Image.Resampling.NEAREST)
plt.figure(figsize=(10, 10)), plt.imshow(im1), plt.show()
```

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">As you can see, it contains some black patches/artifacts and patterns that were not present in the original image—this effect is called aliasing.<br><br>
 Aliasing happens typically because the sampling rate is lower (we had too few pixels!) than
 the Nyquist rate (so one way to avoid aliasing is to increase the sampling rate
 above the Nyquist rate.</p></div>

 <p style="font-family:Bookman Old Style;font-size:24px">Down-sampling and anti-aliasing</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">The problem here is that a single pixel in the output image corresponds to 25 pixels in the
 input image, but we are sampling the value of a single pixel instead. We should be
 averaging over a small area in the input image. This can be done using ANTIALIAS (a high
quality down-sampling filter)<br>
NOTE:<br>
The Image.ANTIALIAS attribute has been removed in recent versions of the PIL library (specifically Pillow, the modern fork of PIL) starting from version 10.0. <br>To fix the issue, you should use Image.Resampling.LANCZOS instead of Image.ANTIALIAS</p></div>

```python
from PIL import Image
import matplotlib.pyplot as plt
im = Image.open("../images-videos/lenna.jpg")
plt.imshow(im), plt.show()
im1 = im.resize((im.width//5, im.height//5), Image.Resampling.LANCZOS)
plt.figure(figsize=(10, 10)), plt.imshow(im1), plt.show()
```

 <div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">Anti-aliasing is generally done by smoothing an image (via convolution of the image with a low-pass filter such as a Gaussian filter) before down-sampling.</p></div>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">now we can use the scikit-image transform module's rescale() function with anti
aliasing to overcome the aliasing problem for another image</p></div>

```python
from skimage.io import imread, imshow, show
from skimage.transform import rescale
import matplotlib.pyplot as plt

img = imread('../images-videos/clock1.jpg')
img1 = img.copy()
plt.figure(figsize=(15, 15))
for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(img1)
    plt.title('iamge size = ' + str(img1.shape[1]) + ' x ' + str(img1.shape[0]), size=18)
    img1 = rescale(img1, scale=0.5, channel_axis=-1, anti_aliasing=False)
plt.show()
```
