<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Import Libraries</p>

```python
import sys
sys.path.append('D:\\Image-Processing-Masterclass\\images-videos')
import createImg
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Initial Variables</p>

```python
h = 200
w = 200
angle = -45
r = 20
```

<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Apply Fourier Transformation</p>

```python
def apply_dft(gray_img):
    dft = cv2.dft(np.float32(gray_img), flags=cv2.DFT_COMPLEX_OUTPUT)
    dft_shift = np.fft.fftshift(dft)
    epsilon = 1e-10
    magnitude_spectrum = 20 * (np.log(cv2.magnitude(dft_shift[:, :, 0], dft_shift[:, :, 1]) + epsilon))
    magnitude_spectrum = cv2.normalize(magnitude_spectrum, None, 0, 255, cv2.NORM_MINMAX)
    return np.uint8(magnitude_spectrum)
```

<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Apply DFT on vertical zeros</p>

```python
img1 = createImg.v_sin_zeros()
img01 = apply_dft(img1)

plt.figure(figsize=(10, 10))

plt.subplot(1, 2, 1)
plt.imshow(img1, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img01, cmap='gray')
plt.axis('off')
plt.show()
```

<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Apply DFT on horizontal zeros</p>

```python
img2 = createImg.h_sin_zeros()
img02 = apply_dft(img2)

plt.figure(figsize=(10, 10))

plt.subplot(1, 2, 1)
plt.imshow(img2, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img02, cmap='gray')
plt.axis('off')
plt.show()
```
