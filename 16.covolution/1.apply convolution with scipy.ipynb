<p style="font-family:Bookman Old Style;font-size:27px; text-align:center">Convolution</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">Convolution is an operation that operates on two images, one being an input image and the other one being a mask (also called the kernel) as a filter on the input image, producing an output image.<br><br>
 Convolution filtering is used to modify the spatial frequency characteristics of an image. It works by determining the value of a central pixel by adding the weighted values of all of its neighbors together to compute the new value of the pixel in the output image. The pixel values in the output image are computed by traversing the kernel window through the input image</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Import libraries</p>

```python
import scipy as sc
import numpy as np
from skimage.io import imread
from skimage.color import rgb2gray
import matplotlib.pyplot as plt
```

<p style="font-family:Bookman Old Style;font-size:27px">Reading and Converting Image to Grayscale</p>

```python
img = rgb2gray(imread('../images-videos/cameraman.jpeg'))
```

<p style="font-family:Bookman Old Style;font-size:27px">Define kernels</p>

```python
blur_kernel = np.ones((3, 3)) / 9
edge_laplace_kernel = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]])
```

<p style="font-family:Bookman Old Style;font-size:27px">Covolve images with kernels</p>

```python
blurred_img = sc.signal.convolve2d(img, blur_kernel)
edged_img = np.clip(sc.signal.convolve2d(img, edge_laplace_kernel), 0, 1)
```

<p style="font-family:Bookman Old Style;font-size:27px">Plotting the results</p>

```python
plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.imshow(img, cmap='gray')
plt.title('Original Image', size=17)
plt.axis('off')

plt.subplot(132)
plt.imshow(blurred_img, cmap='gray')
plt.title('Box Blur', size=17)
plt.axis('off')

plt.subplot(133)
plt.imshow(edged_img, cmap='gray')
plt.title('Laplace Edge Detection', size=17)
plt.axis('off')
plt.show()
```

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px"> With scipy.convolve2d(), we can sharpen an RGB image as well. We have to apply the convolution separately for each image channel.<br>
 Let's use the palace image, with the emboss kernel and the schar edge detection kernel:</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Importing Required libraries</p>

```python
import scipy as sc
import numpy as np
from skimage.io import imread
import matplotlib.pyplot as plt
```

<p style="font-family:Bookman Old Style;font-size:27px">Loading and Normalizing the Image</p>

```python
img = imread('../images-videos/palace.jpg') / 255
```

<p style="font-family:Bookman Old Style;font-size:27px">Defining Convolution Kernels</p>

```python
emboss_kernel = np.array([[-2, -1, 0], [-1, 1, 1], [0, 1, 2]])
scharr_kernel_X = np.array([[-3, 0, 3], [-10, 0, 10], [-3, 0, 3]])
scharr_kernel_Y = np.array([[3, 10, 3], [0, 0, 0], [-3, -10, -3]])
```

<p style="font-family:Bookman Old Style;font-size:27px">Applying the Emboss Effect</p>

```python
emboss_img = np.zeros(img.shape)

for i in range(3):
    emboss_img[..., i] = np.clip(sc.signal.convolve2d(img[..., i], emboss_kernel,
                                                      mode='same', boundary='symm'), 0, 1)
```

<p style="font-family:Bookman Old Style;font-size:27px">Performing Scharr Edge Detection</p>

```python
edged_img_x = np.zeros(img.shape)
edged_img_y = np.zeros(img.shape)
for i in range(3):
    edged_img_x[:, :, i] = np.clip(sc.signal.convolve2d(img[:, :, i], scharr_kernel_X,
                                                        mode='same', boundary='symm'), 0, 1)
    edged_img_y[:, :, i] = np.clip(sc.signal.convolve2d(img[:, :, i], scharr_kernel_Y,
                                                        mode='same', boundary='symm'), 0, 1)
# Compute the gradient magnitude
edges_img = np.clip(edged_img_x + edged_img_y, 0, 1)
```

<p style="font-family:Bookman Old Style;font-size:27px">Displaying Results</p>

```python
plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.imshow(img)
plt.title('Original Image', size=17)
plt.axis('off')

plt.subplot(132)
plt.imshow(emboss_img)
plt.title('Embossed image', size=17)
plt.axis('off')

plt.subplot(133)
plt.imshow(edges_img)
plt.title('Scharr Edge Detection', size=17)
plt.axis('off')
plt.show()
```

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">In the last code we normalized the image but we can covert it to float32; so this code looks like in the below:</p></div>

```python
import scipy as sc
import numpy as np
from skimage.io import imread
import matplotlib.pyplot as plt

img = imread('../images-videos/palace.jpg').astype(np.float32)

emboss_kernel = np.array([[-2, -1, 0], [-1, 1, 1], [0, 1, 2]])
scharr_kernel_X = np.array([[-3, 0, 3], [-10, 0, 10], [-3, 0, 3]])
scharr_kernel_Y = np.array([[3, 10, 3], [0, 0, 0], [-3, -10, -3]])

emboss_img = np.zeros(img.shape)

for i in range(3):
    emboss_img[..., i] = np.clip(sc.signal.convolve2d(img[..., i], emboss_kernel,
                                                      mode='same', boundary='symm'), 0, 255)

edged_img_x = np.zeros(img.shape)
edged_img_y = np.zeros(img.shape)
for i in range(3):
    edged_img_x[:, :, i] = np.clip(sc.signal.convolve2d(img[:, :, i], scharr_kernel_X,
                                                        mode='same', boundary='symm'), 0, 255)
    edged_img_y[:, :, i] = np.clip(sc.signal.convolve2d(img[:, :, i], scharr_kernel_Y,
                                                        mode='same', boundary='symm'), 0, 255)
edges_img = np.clip(edged_img_x + edged_img_y, 0, 255)
plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.imshow(img.astype(np.uint8))
plt.title('Original Image', size=17)
plt.axis('off')

plt.subplot(132)
plt.imshow(emboss_img.astype(np.uint8))
plt.title('Embossed image', size=17)
plt.axis('off')

plt.subplot(133)
plt.imshow(edges_img.astype(np.uint8))
plt.title('Scharr Edge Detection', size=17)
plt.axis('off')
plt.show()
```

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">With scipy.ndimage.convolve(), we can sharpen an RGB image directly (we do not have to apply the convolution separately for each image channel).<br>
 Use the (( )) image with the sharpen kernel:</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Importing Required libraries</p>

```python
import scipy as sc
import numpy as np
from skimage.io import imread
import matplotlib.pyplot as plt
```

<p style="font-family:Bookman Old Style;font-size:27px">Loading and Preparing the Image</p>

```python
img = imread('../images-videos/lenna.jpg').astype(np.float32)
```

<p style="font-family:Bookman Old Style;font-size:27px">Defining the Sharpening Kernel</p>

```python
sharpen_kernel = np.array([0, -1, 0, -1, 5, -1, 0, -1, 0]).reshape((3, 3, 1))
```

<p style="font-family:Bookman Old Style;font-size:27px">Applying the Sharpening Filter</p>

```python
sharpened_img = sc.ndimage.convolve(img, sharpen_kernel, mode='nearest')
sharpened_img = np.clip(sharpened_img, 0, 255).astype(np.uint8)
```

<p style="font-family:Bookman Old Style;font-size:27px">Plotting the Images</p>

```python
plt.figure(figsize=(10, 5))

plt.subplot(121)
plt.imshow(img.astype(np.uint8))
plt.title('Original Image', size=17)
plt.axis('off')

plt.subplot(122)
plt.imshow(sharpened_img)
plt.title('Sharpend Image', size=17)
plt.axis('off')
plt.show()
```
