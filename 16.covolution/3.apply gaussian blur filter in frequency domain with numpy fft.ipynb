<p style="font-family:Bookman Old Style;font-size:27px">Import Libraries</p>

```python
import cv2
import numpy as np
from numpy.fft import fft2, fftshift, ifft2
import matplotlib.pyplot as plt
from scipy.signal import gaussian
```

<p style="font-family:Bookman Old Style;font-size:27px">Load and Convert Image to Grayscale</p>

```python
original_image = cv2.imread('../images-videos/lenna.jpg')
grayscale_image = cv2.cvtColor(original_image, cv2.COLOR_BGR2GRAY)
```

<p style="font-family:Bookman Old Style;font-size:27px">Create Gaussian Kernel for Filtering</p>

```python
vertical_kernel = gaussian(grayscale_image.shape[0], 5)
horizontal_kernel = gaussian(grayscale_image.shape[1], 5)
gaussian_kernel = np.outer(vertical_kernel, horizontal_kernel)
```

<p style="font-family:Bookman Old Style;font-size:27px">Compute Frequency Representation of the Gaussian Kernel</p>

```python
kernel_frequency_domain = fft2(gaussian_kernel)
kernel_magnitude_spectrum = fftshift(np.abs(kernel_frequency_domain))
```

<p style="font-family:Bookman Old Style;font-size:27px">Compute Frequency Representation of the Grayscale Image</p>

```python
image_frequency_domain = fft2(grayscale_image)
shifted_image_frequency = fftshift(image_frequency_domain)
image_magnitude_spectrum = 20 * np.log1p(np.abs(shifted_image_frequency) + 1)
```

<p style="font-family:Bookman Old Style;font-size:27px">Compute Frequency Representation of the Shifted Gaussian Kernel</p>

```python
shifted_kernel_frequency = fft2(fftshift(gaussian_kernel))
```

<p style="font-family:Bookman Old Style;font-size:27px">Perform Frequency Domain Convolution (Multiplying Image Frequency with Kernel Frequency)</p>

```python
convolved_frequency_domain = image_frequency_domain * shifted_kernel_frequency
```

<p style="font-family:Bookman Old Style;font-size:27px">Convert Back to Spatial Domain</p>

```python
convolved_image = np.abs(ifft2(convolved_frequency_domain))
```

<p style="font-family:Bookman Old Style;font-size:27px">Display results</p>

```python
plt.figure(figsize=(10, 6))
plt.subplot(231)
plt.imshow(grayscale_image, cmap='gray')
plt.title('original image', size=14)
plt.axis('off')

plt.subplot(232)
plt.imshow(kernel_magnitude_spectrum, cmap='gray')
plt.title('gaussian kernel', size=14)
plt.axis('off')

plt.subplot(233)
plt.imshow(convolved_image, cmap='gray')
plt.title('output image', size=14)
plt.axis('off')

plt.subplot(234)
plt.imshow(image_magnitude_spectrum, cmap='gray')
plt.title('original image spectrum', size=14)
plt.axis('off')

plt.subplot(235)
plt.imshow(20 * np.log1p(kernel_magnitude_spectrum), cmap='gray')
plt.title('gaussian kernel spectrum', size=14)
plt.axis('off')

plt.subplot(236)
plt.imshow(20 * np.log1p(fftshift(np.abs(convolved_frequency_domain))), cmap='gray')
plt.title('output image spectrum', size=14)
plt.axis('off')
plt.show()
```
