<p style="font-family:Bookman Old Style;font-size:27px">Bitwise Operations</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">Bitwise operations are operations that directly manipulate individual bits of binary numbers. They are fundamental in low-level programming and are often used in scenarios where performance is critical, like in embedded systems, cryptography, data compression, graphics, and network protocols. Here’s an overview of key bitwise operations and concepts:</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Binary Representation</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">- Computers represent data in binary, a base-2 numeral system, where each bit can be either 0 or 1.
<br><br>- Bitwise operations work directly on this binary data at the bit level, making them fast and efficient for specific tasks.</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Common Bitwise Operations</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">AND (&): Sets each bit to 1 if both corresponding bits of the operands are 1, otherwise 0<br>
&nbsp;&nbsp;&nbsp; - Example: 5 & 3 where 5 is 0101 and 3 is 0011, results in 0001 (or 1)<br><br>
OR (|): Sets each bit to 1 if at least one of the corresponding bits of the operands is 1<br>
&nbsp;&nbsp;&nbsp; - Example: 5 | 3 gives 0111 (or 7)<br><br>
XOR (^): Sets each bit to 1 only if the corresponding bits of the operands are different<br>
&nbsp;&nbsp;&nbsp; - Example: 5 ^ 3 gives 0110 (or 6)<br><br>
NOT (~): Flips all bits in a number, converting 0s to 1s and vice versa<br>
&nbsp;&nbsp;&nbsp; - Example: ~5 flips 0101 to 1010 (in two’s complement, this is often interpreted as -6)</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Useful Applications of Bitwise operations</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">
- Memory Efficiency: Used in data compression and to reduce storage.<br>
- Encryption: XOR and bitwise manipulations are common in encryption algorithms.<br>
- Flags and Permissions: Used to set, clear, and check permissions or configuration options.<br>
- Graphics Processing: Efficient manipulation of pixel data and color values.</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Bitwise Operations on Images</p>

<div style="margin:0 130px 0 30px;text-align:justify"><p style="font-family:Source Sans Pro;font-size:20px">Bitwise operations can be applied to images to manipulate pixel values at the binary level, creating effects or filters, performing masking, and working with image channels. In digital images, each pixel's color is represented in binary (usually as a combination of red, green, and blue values, each stored as 8 bits, giving a range of 0–255 per color channel)</p></div>

<p style="font-family:Bookman Old Style;font-size:27px">Show bitwise on square and circle image</p>

```python
import cv2
import numpy as np

# make a square and a circle in zeros
window1 = np.zeros((400,400),dtype='uint8')
window2 = np.zeros((400,400),dtype='uint8')

square = cv2.rectangle(window1,(160,180),(360,360),(255,255,255),-1)
circle = cv2.circle(window2,(200,200),130,(255,255,255),-1)

cv2.imshow('square',square)
cv2.imshow('circle',circle)

# apply bitwise NOT on a square and a circle
window3 = np.zeros((400,400),dtype='uint8')
square_NOT = cv2.bitwise_not(square)
cv2.imshow('NOT square',square_NOT)

window4 = np.zeros((400,400),dtype='uint8')
circle_NOT = cv2.bitwise_not(circle)
cv2.imshow('NOT circle',circle_NOT)

# apply bitwise AND on a square and a circle
window5 = np.zeros((400,400),dtype='uint8')
img_AND = cv2.bitwise_and(square,circle)
cv2.imshow('and image',img_AND)

# apply bitwise OR on a square and a circle
window6 = np.zeros((400,400),dtype='uint8')
img_OR = cv2.bitwise_or(square,circle)
cv2.imshow('or image',img_OR)

# apply bitwise XOR on a square and a circle
window7 = np.zeros((400,400),dtype='uint8')
img_XOR = cv2.bitwise_xor(square,circle)
cv2.imshow('xor image',img_XOR)

while True:
    if cv2.waitKey() == ord('q'):
        break
cv2.destroyAllWindows()
```

<p style="font-family:Bookman Old Style;font-size:27px">Show concept of bitwise on 2 images</p>

```python
import cv2

img1 = cv2.imread('../images-videos/phone.jpg')
img2 = cv2.imread('../images-videos/sea.jpg')

cv2.imshow('original image 1',img1)
cv2.imshow('original image 2',img2)

img1 = cv2.resize(img1,(250,250))
img2 = cv2.resize(img2,(250,250))

# work on an image...the base image here is girl250.jpg
grayIMG1 = cv2.cvtColor(img1,cv2.COLOR_BGR2GRAY)
thresholdValue1, imageThreshold1 = cv2.threshold(grayIMG1,100,255,cv2.THRESH_BINARY)
not_img = cv2.bitwise_not(imageThreshold1)
# bitwise operations
img_AND = cv2.bitwise_and(img1,img2,mask=not_img)
img_OR = cv2.bitwise_or(img1,img2,mask=not_img)
img_XOR = cv2.bitwise_xor(img1,img2,mask=not_img)

# show bitwise images
cv2.imshow('threshold',imageThreshold1)
cv2.imshow('NOT img',not_img)

cv2.imshow('AND',img_AND)
cv2.imshow('OR',img_OR)
cv2.imshow('XOR',img_XOR)

while True:
    if cv2.waitKey() == ord('q'):
        break
cv2.destroyAllWindows()
```
